#program the count for vowels and consonats in a string

```python
a=str(input())
b=a.lower()
vowels = 'aeiou'
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print(count_v)
print(count_c)
```
