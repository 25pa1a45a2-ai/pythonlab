# program that checks that given number is palindrome or not using function

```python
def palindrome(num):
 x=num
 rev=0
 while(num!=0):
  rem=num%10
  rev=rev*10+rem
  num=num//10
 if rev==x:
  return True
 else:
  return False
print(palindrome(1234))
```

#program that reverse a given string using functions

```python
def reverse(name):
  rev=name[::-1]
  return rev
print(reverse("bala"))
```

```python
def reverse_str(n):
  rev=""
  for i in range(len(n)-1,-1,-1):
    rev=rev+n[i]
  print(rev)
a=str(input())
reverse_str(a)
```

#program to print the multiple values

```python
def ml(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
rs,rc,ravg=ml(1,2,3,4,5)
print(f"the sum is{rs} the length is{rc} the average{int(ravg)}")
```
