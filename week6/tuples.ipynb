#TUPLES

#creation of tuples using operation

```python
t=(1,2,3,4,5)
print(t)
```

#operation in tuple

```python
t1=(1,2,3,4,5)
t2=(2,5,3,76,3)
print(t1+t2)
print(t1*2)
if 3 in t1:
  print("yes")
else:
  print("no")
print(t2[::-1]) #reversing
print(t1[2:4]) #slicing
```

#functions in tuple

```python
t1=(3,5,3,1,6,7)
print(sum(t1))
print(max(t1))
print(min(t1))
print(len(t1))
```

#methods in tuples

```python
t3=(34,54,23,54)
print(t3.count(54))
print(t3.index(23))
```
